```python
import litellm
import gradio as gr
from  smolagents import CodeAgent,DuckDuckGoSearchTool,LiteLLMModel
import google.generativeai as genai
```

```python
# enable the debug option, useful for troubleshooting
litellm._turn_on_debug()
```

```python
genai.configure(api_key=os.getenv("GEMINI_API_KEY"))

# initialize the google model for SmolAgents
model = LiteLLMModel(
    model_id=os.getenv("GEMINI_MODEL"),
   api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
ml_code_agent = CodeAgent(tools=[DuckDuckGoSearchTool()],
                  additional_authorized_imports=['pandas','numpy','sklearn','json'],
                  model=model, 
                  stream_outputs=True)

# agent.run("what is the distance between the earth and moon?")
```

```python
# Function to Generate ML Code Step-by-Step

def generate_ml_code(prompt):
    full_prompt = """
    create a step-by-step plan to write python code for svm model using sklearn to
    classify the iris dataset.
    """
    try:
        response = ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"Error: {str(e)}"
    
```

```python
import gradio as gr
iface = gr.Interface(
    fn = generate_ml_code,
    inputs="text",
    outputs="text",
    title = "ML code generator",
    description="Enter a prompt to generate required use information"
)
```

```python
iface.launch(share =True)
```
