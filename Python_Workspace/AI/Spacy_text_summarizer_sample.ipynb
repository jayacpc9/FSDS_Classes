```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation

stopwords = list(STOP_WORDS)
print("Stop words length = ",len(stopwords))
stopwords

```

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
doc
```

```python
tokens = [tokens.text for tokens in doc]
print(tokens)
```

# get the word frequencies in the given text paragraph

```python

word_frequencies={}

for word in doc:
  if word.text.lower() not in stopwords:
    if word.text.lower() not in punctuation:
      if word.text not in word_frequencies.keys():
          word_frequencies[word.text] = 1
      else:
        word_frequencies[word.text]+=1

print('Length of word frequencies = ',len(word_frequencies))
word_frequencies
```

len

```python
max_frequency = max(word_frequencies.values())
max_frequency
```

# to get normalized/weighted frequencies you should devide all frequencies with max_frequency
# in our case divide by 11


```python
for word in word_frequencies.keys():
    word_frequencies[word] = word_frequencies[word]/max_frequency

word_frequencies
```

# get sentences from the text/paragraph

```python
sentences_tokens = [sentence for sentence in doc.sents]
print("Total no of Senctences = ",len(sentences_tokens))
sentences_tokens
```

# we are going to calculate the sentence score, to calculate the sentence score 
# add the word_frequency score for all words in the sentence, do this for each sentence.



```python
sentence_scores = {}

for sent in sentence_tokens:
    for word in sent:
        if word.text.lower() in word_frequencies.keys():
            if sent not in sentence_scores.keys():
                sentence_scores[sent] = word_frequencies[word.text.lower()]
            else:
                sentence_scores[sent] += word_frequencies[word.text.lower()]

print("len(sentence_scores) = ",len(sentence_scores))
sentence_scores

```

# lets say our case study was 30% sentence with maximum scores
# 30% of 15 is 4.5, so consider 4 as a whole number

```python
from heapq import nlargest 

print("len(sentences_tokens) = ",len(sentences_tokens))
select_length = int(len(sentences_tokens)*.3)
select_length
```

# we have to select maximum 4 sentences out of all sentences 


```python
summary = nlargest(select_length,sentence_scores, key = sentence_scores.get)
print("len(summary) = ",len(summary))
summary
```

# if i need to combine these top 3 sentencs then 


```python
final_summary = [sent for sent in summary]
final_summary
```
