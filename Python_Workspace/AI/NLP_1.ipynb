# Types of Tokens
#### Bigrams: tokens with 2 words
#### Trigrams : tokens with 3 words
#### Ngrams : tokens of any number of words


```python
from nltk.util import bigrams,trigrams, ngrams
string = "generative ai & agentic ai is automate the industry & more focus on every domain"
quotes_tokens = nltk.word_tokenize(string)

print("Length of quotes_tokens = ",len(quotes_tokens))
print(quotes_tokens)
```

```python
quotes_bigram = list(nltk.bigrams(quotes_tokens))

print("Length of quotes_bigram =",len(quotes_bigram))
quotes_bigram
```

```python
quotes_trigram = list(nltk.trigrams(quotes_tokens))
print("Length of quotes_trigram =",len(quotes_trigram))
quotes_trigram
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens,n=5))
print("Length of quotes_ngrams =",len(quotes_ngrams))
quotes_ngrams
```

# Stemming and Lematization
#### Stemming : Root form of a word.  It gives a root word(something like a dictionary)

#### Lematization : Complete word.
#### for exampe : if token = Playing then 
#### Stemming= play, Lematization = playing

```python
from nltk.stem import PorterStemmer
pst = PorterStemmer()
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
words_to_stem = ['give','giving','given','gave','thinking','loving','maximum','Hello']
for words in words_to_stem:
    print(words+" : "+pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()

for words in words_to_stem:
    print(words+" : "+lst.stem(words))
```

# SnowballStemmer : very useful for languages, pass language

```python
from nltk.stem import SnowballStemmer
def demo_show_ball_stemmer(language):
    sbst = SnowballStemmer(language=language)
    for words in words_to_stem:
        print(words+" : "+sbst.stem(words))
    print("*"*50)

demo_show_ball_stemmer('english')
```

```python
demo_show_ball_stemmer('german')
demo_show_ball_stemmer('french')
```

```python
import nltk
from nltk import pos_tag, word_tokenize, RegexpParser

# Download necessary NLTK data files (only need to do this once)
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

# Sample text
text = "full stack datasceince, generative ai, agenti ai, llm model keep increase by different compnay"

# Tokenize the text
tokens = word_tokenize(text)

# Perform part-of-speech tagging
tagged_tokens = pos_tag(tokens)

# Define a chunk grammar
chunk_grammar = r"""
  NP: {<DT>?<JJ>*<NN>}   # Noun Phrase
  VP: {<VB.*><NP|PP>*}    # Verb Phrase
  PP: {<IN><NP>}          # Prepositional Phrase
"""

# Create a chunk parser
chunk_parser = RegexpParser(chunk_grammar)

# Parse the tagged tokens
chunked = chunk_parser.parse(tagged_tokens)

# Print the chunked output
print(chunked)

# Optionally, you can visualize the chunks
chunked.draw()
```
