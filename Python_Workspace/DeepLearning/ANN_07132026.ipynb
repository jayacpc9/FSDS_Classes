```python
import os

folder_path =os.getcwd()
file_name='/Churn_Modelling.csv'

full_file_path = folder_path+file_name
print('Full File Path = ',full_file_path)
```

```python
dataset = pd.read_csv(f""+full_file_path)
dataset.columns
```

```python
dataset.head()
```

```python
X = dataset.iloc[:,3:-1]
y = dataset.iloc[:,-1]

X
```

```python
X.columns
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
# X[:,2] = le.fit_transform(X[:,2])
X['Gender'] = le.fit_transform(X['Gender'])
```

```python
X.head()
```

```python
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])],remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
print("X Train = ",X_train.shape)
print("X Test = ",X_test.shape)
print("Y Train = ",y_train.shape)
print("Y Test = ",y_test.shape)
```

# ANN Model Building Architecture

```python
ann = tf.keras.models.Sequential() # establish layer one after another 
```

```python
ann.add(tf.keras.layers.Dense(units=6)) # input layer, densly connect layers to neurons (6 neurons)
```

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu')) # hidden layer 1
```

```python
# ann.add(tf.keras.layers.Dense(units=6, activation='relu')) # hidden layer 2
```

```python
# ann.add(tf.keras.layers.Dense(units=6, activation='relu')) # hidden layer 3
```

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid')) # output layer 1
```

```python
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(X_train,y_train,batch_size=32,epochs=100) # taking 32 records as one batch and epochs for 10 times
```

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred > 0.5)

# Ensure y_test is a numpy array and reshape both to 2D column vectors
y_pred_reshaped = np.array(y_pred).reshape(-1, 1)
y_test_reshaped = np.array(y_test).reshape(-1, 1)

# Concatenate side-by-side: [Predicted, Actual]
comparison = np.concatenate((y_pred_reshaped, y_test_reshaped), axis=1)

# Print first 10 rows to inspect
print("Predicted | Actual")
print(comparison[:10])
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
from sklearn.metrics import accuracy_score
ac = accuracy_score(y_test,y_pred)
print(ac)
```
